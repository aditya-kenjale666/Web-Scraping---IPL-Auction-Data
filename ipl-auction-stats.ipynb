```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = 'https://www.iplt20.com/news/article/tata-ipl-2025-player-retentions-list-announced'

r = requests.get(url)

soup = BeautifulSoup(r.text, 'html.parser')

# Find table
table = soup.find("table", class_="min-w-full border border-gray-300")

# First row = header
first_row = table.find("tr")

header = []

for i in first_row.find_all("td"):
    header.append(i.get_text(" ", strip=True))

# Create DataFrame
df = pd.DataFrame(columns=header)

# Find all rows
rows = table.find_all("tr")

for i in rows[1:]:
    data = i.find_all("td")

    row = [td.get_text(" ", strip=True) for td in data]

    df.loc[len(df)] = row

print(df)

df.to_csv("TATA_IPL_2025_Player_Retentions.csv", index=False)
```
